## 2. Modules and imports

## 4. Exceptions and tracebacks

An **exception** is an error detected while code is running. Python displays a traceback that helps identify where the error occurred and what type of exception was raised.

Common examples:
- `ValueError`: a value has the right general type but invalid content.
- `TypeError`: an operation is used with an incompatible type.
- `FileNotFoundError`: a requested file does not exist.
- `ZeroDivisionError`: a number is divided by zero.
- `KeyError`: a dictionary key is missing.

A traceback is useful evidence. Read the final line for the exception type and message, then inspect the relevant line of code.

````python
# A safe example: inspect a conversion without stopping the notebook.
quantity_text = "three"

try:
    quantity = int(quantity_text)
    print(f"Quantity: {quantity}")
except ValueError as error:
    print(f"Could not convert quantity: {error}")
````

**Expected output**
```text
Could not convert quantity: invalid literal for int() with base 10: 'three'
```

The input is text, but it does not contain a valid integer. Catching `ValueError` lets the program show a useful message and continue.

## 5. `try`, `except`, `else`, and `finally`

Use `try` for code that may raise a known exception. Use `except` to handle that exception. The optional `else` block runs only if the `try` block succeeds. The optional `finally` block runs whether an exception occurred or not.

Catch specific exceptions whenever possible. A broad `except:` can hide programming mistakes and make debugging harder.

````python
quantity_text = "3"

try:
    quantity = int(quantity_text)
except ValueError:
    print("Quantity must be a whole number.")
else:
    print(f"Accepted quantity: {quantity}")
finally:
    print("Quantity validation finished.")
````

**Expected output**
```text
Accepted quantity: 3
Quantity validation finished.
```

Try changing `quantity_text` to `"three"` and run the cell again. Notice that the `except` branch runs, while `else` does not.

### A business example: validating a quantity from text

Input from a form or file often arrives as text. Convert it carefully and give a clear message when the value is invalid.

````python
quantity_text = "2"
unit_price = 24.99

try:
    quantity = int(quantity_text)
    if quantity < 0:
        raise ValueError("Quantity cannot be negative.")
except ValueError as error:
    print(f"Invalid order quantity: {error}")
else:
    line_total = quantity * unit_price
    print(f"Line total: ${line_total:.2f}")
````

## 6. Raising exceptions and validating inputs

Sometimes a function should reject invalid data instead of silently returning a misleading result. Use `raise` to signal a problem with a clear exception type and message.

The validation rules below are simple teaching rules for AdventureWorks. A real system should document its own business rules.

````python
def calculate_line_total(unit_price, quantity):
    """Return a line total after validating basic inputs."""
    if unit_price < 0:
        raise ValueError("unit_price cannot be negative")
    if quantity < 0:
        raise ValueError("quantity cannot be negative")
    return unit_price * quantity

print(f"Valid line total: ${calculate_line_total(24.99, 2):.2f}")
````

**Expected output**
```text
Valid line total: $49.98
```

The next cell demonstrates the invalid case safely by catching the expected exception.

````python
try:
    calculate_line_total(-5.00, 2)
except ValueError as error:
    print(f"Order validation failed: {error}")
````

**Why this matters:** An invalid price should be identified at the point where the rule is checked. This makes the function easier to reuse and test.

## 7. Reading and writing text files

Files let a program keep information after a notebook session ends. Python's `open()` function can read or write a file.

Common modes:
- `"r"`: read an existing file.
- `"w"`: write a file, replacing its previous content if it exists.
- `"a"`: append to an existing file or create it.
- `"x"`: create a new file and fail if it already exists.

Use `with open(...)` so the file is closed automatically, even if an error occurs. Specify `encoding="utf-8"` for text files unless another encoding is required.

````python
from pathlib import Path

data_folder = Path("adventureworks_data")
data_folder.mkdir(exist_ok=True)
summary_path = data_folder / "order_summary.txt"

with open(summary_path, "w", encoding="utf-8") as file:
    file.write("AdventureWorks Order Summary\n")
    file.write("Order ID: AW-3105\n")
    file.write("Customer: Lina Haddad\n")
    file.write("Status: Processing\n")

print(f"Saved file: {summary_path}")
````

The `Path` object builds a path in a readable way. `mkdir(exist_ok=True)` creates the folder if needed and does not fail if it already exists.

### Reading the file

Use `"r"` to read the text that was saved.

````python
with open(summary_path, "r", encoding="utf-8") as file:
    saved_text = file.read()

print(saved_text)
````

**Expected output**
```text
AdventureWorks Order Summary
Order ID: AW-3105
Customer: Lina Haddad
Status: Processing
```

### Appending to a file

Append mode adds content to the end instead of replacing the existing file.

````python
with open(summary_path, "a", encoding="utf-8") as file:
    file.write("Review note: Confirm inventory before shipping.\n")

with open(summary_path, "r", encoding="utf-8") as file:
    print(file.read())
````

**Important:** Running a `"w"` cell again replaces the file contents. Use `"a"` when you intend to add a new line.

## 8. File paths and safe file handling

A file path identifies a file location. Relative paths are interpreted from the current working directory, which can vary between environments.

Use `pathlib.Path` to build paths and check whether a file exists. Handle a missing file with a specific exception when it is an expected possibility.

````python
from pathlib import Path

missing_path = Path("adventureworks_data/missing_report.txt")

try:
    with open(missing_path, "r", encoding="utf-8") as file:
        content = file.read()
except FileNotFoundError:
    print(f"Report not found: {missing_path}")
else:
    print(content)
````

**Expected output**
```text
Report not found: adventureworks_data/missing_report.txt
```

Do not catch every possible exception just to make an error disappear. Handle errors that you can explain and respond to appropriately.

## 10. Integrated AdventureWorks workflow

Now combine functions, validation, exceptions, text output, and JSON. The workflow calculates an order subtotal, creates a summary, and saves the original order record as JSON.

````python
def calculate_order_subtotal(order_record):
    """Return the sum of line totals in an order record."""
    subtotal = 0.0
    for item in order_record["items"]:
        subtotal += calculate_line_total(item["unit_price"], item["quantity"])
    return subtotal


def create_order_summary(order_record):
    """Return a readable order summary as text."""
    subtotal = calculate_order_subtotal(order_record)
    return (
        f"Order: {order_record['order_id']}\n"
        f"Customer: {order_record['customer']}\n"
        f"Status: {order_record['status']}\n"
        f"Subtotal: ${subtotal:.2f}"
    )


print(create_order_summary(order))
````

**Expected output**
```text
Order: AW-3105
Customer: Lina Haddad
Status: Processing
Subtotal: $83.98
```

Next, save the summary as a text file and the record as JSON.

````python
summary_path = data_folder / "order_AW-3105_summary.txt"
json_path = data_folder / "order_AW-3105.json"

with open(summary_path, "w", encoding="utf-8") as file:
    file.write(create_order_summary(order))

with open(json_path, "w", encoding="utf-8") as file:
    json.dump(order, file, indent=2)

print(f"Text summary saved: {summary_path}")
print(f"Structured order saved: {json_path}")
````

**Instructor note:** The text file is convenient for a person to read. The JSON file preserves the nested structure for another program to load.

## 12. Instructor answer key

````python
import json
from pathlib import Path

def calculate_validated_subtotal(order_record):
    """Calculate an order subtotal after validating each line."""
    subtotal = 0.0
    for item in order_record["items"]:
        subtotal += calculate_line_total(item["unit_price"], item["quantity"])
    return subtotal


def build_order_report(order_record):
    """Return a text report for an AdventureWorks order."""
    subtotal = calculate_validated_subtotal(order_record)
    return (
        f"Order ID: {order_record['order_id']}\n"
        f"Customer: {order_record['customer']}\n"
        f"Status: {order_record['status']}\n"
        f"Subtotal: ${subtotal:.2f}"
    )


output_folder = Path("adventureworks_data")
output_folder.mkdir(exist_ok=True)
text_path = output_folder / "challenge_report.txt"
json_path = output_folder / "challenge_order.json"

with open(text_path, "w", encoding="utf-8") as file:
    file.write(build_order_report(order))

with open(json_path, "w", encoding="utf-8") as file:
    json.dump(order, file, indent=2)

with open(json_path, "r", encoding="utf-8") as file:
    restored = json.load(file)

assert restored["order_id"] == order["order_id"]
print(build_order_report(restored))
print("JSON verification passed.")
````
