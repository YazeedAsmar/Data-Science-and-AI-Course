## 2. Boolean conditions and comparisons

A condition is an expression that evaluates to `True` or `False`. Conditions allow a program to choose what to do next.

Comparison operators include `==`, `!=`, `>`, `<`, `>=`, and `<=`.

````python
order_total = 149.50

print(order_total > 100)
print(order_total == 149.50)
print(order_total < 50)
````

## 3. `if`, `elif`, and `else`

Use `if` when a block should run only when a condition is true. Use `elif` to check another condition if the earlier condition was false. Use `else` for the remaining cases.

Indentation is part of Python syntax. The statements inside each branch must be indented consistently.

````python
order_total = 149.50

if order_total >= 250:
    order_priority = "High-value review"
elif order_total >= 100:
    order_priority = "Standard review"
else:
    order_priority = "Regular processing"

print(f"Order priority: {order_priority}")
````

**Expected output**
```text
Order priority: Standard review
```

Python checks the conditions from top to bottom and runs the first matching branch. The order of conditions matters: put the most specific or highest threshold first when the rules overlap.

````python
stock_quantity = 8
reorder_level = 10

if stock_quantity == 0:
    stock_status = "Out of stock"
elif stock_quantity <= reorder_level:
    stock_status = "Reorder soon"
else:
    stock_status = "Stock is sufficient"

print(stock_status)
````

## 4. Combining conditions

Use `and` when all conditions must be true, `or` when at least one condition must be true, and `not` to reverse a Boolean condition.

Parentheses can make complex conditions easier to read.

````python
order_total = 320.00
order_status = "Processing"
payment_confirmed = True

requires_manual_review = order_total >= 250 and order_status == "Processing"
can_ship = payment_confirmed and order_status == "Processing"

print(f"Manual review required: {requires_manual_review}")
print(f"Ready to ship: {can_ship}")
````

**Expected output**
```text
Manual review required: True
Ready to ship: True
```

**Important:** These are simplified teaching rules. Real order processing would need more checks, such as inventory availability, payment state, fraud controls, and shipping restrictions.

````python
order_status = "Cancelled"
payment_confirmed = True

can_ship = payment_confirmed and order_status != "Cancelled"
needs_attention = order_status == "Cancelled" or not payment_confirmed

print(f"Can ship: {can_ship}")
print(f"Needs attention: {needs_attention}")
````

## 5. `for` loops

A `for` loop repeats a block for each item in an iterable, such as a list or tuple. Use it when you want to process each item in a collection.

The loop variable takes the value of the current item during each repetition.

````python
featured_products = ["Wireless Mouse", "USB-C Hub", "Desk Lamp"]

for product_name in featured_products:
    print(f"Featured product: {product_name}")
````

### Apply a business rule to each record

A loop can examine each dictionary in a list and apply the same rule to every record.

````python
for product in inventory:
    if product["stock"] <= product["reorder_level"]:
        print(f"Reorder needed: {product['product_name']}")
````

**Expected output**
```text
Reorder needed: Wireless Mouse
Reorder needed: USB-C Hub
```

**Why this matters:** The same rule can be applied to many records without copying the same code for each product.

## 6. `range()`, `enumerate()`, and `zip()`

### `range()`

`range()` produces a sequence of integers, often used when you need to repeat an action a specific number of times. The stop value is excluded.

````python
for attempt in range(1, 4):
    print(f"Checking order status — attempt {attempt}")
````

### `enumerate()`

Use `enumerate()` when you need both an item's position and its value. The optional `start` argument controls the starting counter.

````python
for position, product_name in enumerate(featured_products, start=1):
    print(f"{position}. {product_name}")
````

### `zip()`

Use `zip()` to process corresponding items from two or more iterables together. In this example, each product name is paired with its price. The iterables should have matching lengths when every value must be processed.

````python
product_names = ["Wireless Mouse", "USB-C Hub", "Desk Lamp"]
product_prices = [24.99, 34.00, 28.75]

for product_name, product_price in zip(product_names, product_prices):
    print(f"{product_name}: ${product_price:.2f}")
````

## 7. Iterating through dictionaries and nested data

A dictionary stores key-value pairs. Use `.items()` when you need both a key and its value, `.keys()` for keys, and `.values()` for values.

````python
campaign = {
    "campaign_id": "PROMO-17",
    "name": "Autumn Essentials",
    "status": "Active"
}

for key, value in campaign.items():
    print(f"{key}: {value}")
````

### Nested records

A list can contain dictionaries, and a dictionary can contain a list. Access each level clearly and use meaningful variable names.

````python
order = {
    "order_id": "AW-2087",
    "customer": "Lina Haddad",
    "items": [
        {"product_name": "Wireless Mouse", "quantity": 2, "unit_price": 24.99},
        {"product_name": "USB-C Hub", "quantity": 1, "unit_price": 34.00}
    ]
}

for item in order["items"]:
    line_total = item["quantity"] * item["unit_price"]
    print(f"{item['product_name']}: ${line_total:.2f}")
````

## 8. `while` loops

A `while` loop repeats as long as its condition is true. Use it when the number of repetitions depends on a condition rather than a fixed collection.

A `while` loop must make progress toward ending. If its condition never becomes false, it can run indefinitely.

````python
remaining_units = 3

while remaining_units > 0:
    print(f"Units remaining to pack: {remaining_units}")
    remaining_units -= 1

print("Packing check complete.")
````

The statement `remaining_units -= 1` decreases the value on each iteration. When the value reaches zero, the condition becomes false and the loop ends.

### A controlled retry example

This example simulates a limited number of checks without requiring interactive input.

````python
checks_remaining = 3
shipment_ready = False

while checks_remaining > 0 and not shipment_ready:
    print(f"Checking shipment readiness. Checks left: {checks_remaining}")
    checks_remaining -= 1

print("Check limit reached." if not shipment_ready else "Shipment is ready.")
````

**Common mistake:** Forgetting to update a variable used in the condition can create an infinite loop. Always identify what changes on each iteration and what makes the loop stop.

## 9. `break` and `continue`

`break` ends the nearest loop immediately. `continue` skips the rest of the current iteration and moves to the next item.

Use these statements only when they make the logic clearer.

````python
for order in orders:
    if order["status"] == "Cancelled":
        print(f"Skipping cancelled order: {order['order_id']}")
        continue

    print(f"Reviewing order: {order['order_id']}")
````

### Stop after finding a target

Use `break` when you have found the item you need and no further iteration is required.

````python
target_order_id = "AW-1044"

for order in orders:
    if order["order_id"] == target_order_id:
        print(f"Found order: {order['order_id']} — {order['status']}")
        break
````

## 10. Common mistakes and debugging

### Mistake 1: Off-by-one errors

The stop value in `range(start, stop)` is excluded.

````python
print(list(range(1, 4)))  # Includes 1, 2, 3 — not 4
````

### Mistake 2: Incorrect indentation

Python uses indentation to identify the body of a condition or loop. The following is incorrect and will cause `IndentationError` if uncommented.

````python
# Intentionally incorrect example — uncomment to observe the error:
# if True:
# print("This line must be indented")
````

### Mistake 3: An infinite `while` loop

If the condition variable never changes, the loop may never stop. In a notebook, interrupt execution if a loop runs unexpectedly.

### Debugging checklist

- Check whether the condition evaluates to the value you expect.
- Confirm indentation inside each branch or loop.
- Check loop boundaries and the stop value in `range()`.
- For `while`, identify the update that makes the condition false.
- Print a small, useful value when you need to inspect progress.
- Test business rules with normal, boundary, and exceptional cases.
