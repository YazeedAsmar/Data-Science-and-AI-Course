## 2. Why NumPy and Pandas?

- **NumPy** provides efficient arrays and numerical operations.
- **Pandas** provides labeled tabular data structures, especially `Series` and `DataFrame`.

Python lists are flexible general-purpose containers. NumPy arrays are designed for numerical operations on arrays of compatible values. Pandas makes it easier to explore and transform tables with named columns.

````python
sample_quantities = [2, 1, 2, 1]
quantity_array = np.array(sample_quantities)

print(f"Python list: {sample_quantities}")
print(f"NumPy array: {quantity_array}")
print(f"Shape: {quantity_array.shape}")
print(f"Dimensions: {quantity_array.ndim}")
print(f"Data type: {quantity_array.dtype}")
````

**Expected output (the exact integer dtype may vary by platform):**
```text
Python list: [2, 1, 2, 1]
NumPy array: [2 1 2 1]
Shape: (4,)
Dimensions: 1
```

The array contains four elements and has one dimension. NumPy arrays usually contain values of a compatible data type.

## 3. NumPy arrays

Create arrays with `np.array()`. Inspect `.shape`, `.ndim`, `.size`, and `.dtype` to understand the data's structure.

````python
quantities = np.array([2, 1, 2, 1])
unit_prices = np.array([24.99, 79.50, 28.75, 34.00])

print(f"Quantity shape: {quantities.shape}")
print(f"Number of dimensions: {quantities.ndim}")
print(f"Number of values: {quantities.size}")
print(f"Price dtype: {unit_prices.dtype}")
````

### Vectorized calculations

Vectorization applies an operation across an array without writing an explicit Python loop. Corresponding positions should refer to the same business record.

````python
line_totals = quantities * unit_prices

print(line_totals)
print(f"Total line value: ${line_totals.sum():.2f}")
print(f"Average line value: ${line_totals.mean():.2f}")
````

**Expected output**
```text
[49.98 79.5  57.5  34.  ]
Total line value: $220.98
```

### Indexing, slicing, and Boolean filtering

NumPy arrays support indexing and slicing. A Boolean condition can select only the values that meet a rule.

````python
print(f"First line total: ${line_totals[0]:.2f}")
print(f"First two totals: {line_totals[:2]}")
print(f"Totals above $50: {line_totals[line_totals > 50]}")
````

## 6. Selecting columns and rows

Select one column with `orders["product"]` and several columns with a list of names: `orders[["order_id", "product"]]`.

- `.loc` selects using index labels.
- `.iloc` selects using integer positions.

The default index begins at zero, but labels and positions are not always the same.

````python
display(orders[["order_id", "product", "status"]].head())

print("First row by position:")
display(orders.iloc[0])

print("Index labels 1 through 3:")
display(orders.loc[1:3, ["order_id", "customer", "status"]])
````

**Important:** `.loc` uses label-based slicing and includes the ending label when present. `.iloc` uses integer positions and excludes the stop position, like ordinary Python slicing.

## 7. Filtering rows

Filtering selects records that meet a condition. Use parentheses around each condition when combining them with `&` (AND) or `|` (OR). Do not use Python's `and` or `or` directly between Pandas Series.

````python
processing_orders = orders[orders["status"] == "Processing"]
display(processing_orders[["order_id", "customer", "product", "status"]])
````

````python
premium_processing = orders[
    (orders["status"] == "Processing") &
    (orders["unit_price"] >= 50)
]
display(premium_processing[["order_id", "product", "unit_price", "status"]])
````

````python
selected_categories = orders[orders["category"].isin(["Home", "Electronics"])]
display(selected_categories[["order_id", "category", "product"]])
````

## 8. Creating and transforming columns

A new column can be calculated from existing columns. This is a common data-preparation operation.

Here, `line_total` is a simplified order-line value before discounts, tax, returns, and other adjustments.

````python
orders["line_total"] = orders["quantity"] * orders["unit_price"]
display(orders[["order_id", "product", "quantity", "unit_price", "line_total"]])
````

Use `np.where(condition, value_if_true, value_if_false)` for a simple vectorized two-way rule.

````python
orders["value_band"] = np.where(
    orders["line_total"] >= 100,
    "High line value",
    "Standard line value"
)
display(orders[["order_id", "line_total", "value_band"]])
````

**Business note:** `value_band` describes a line, not the full order. Clear column names reduce ambiguity.

## 9. Missing values and duplicates

Real datasets can contain missing or repeated records. First identify the issue, then choose a handling strategy based on business meaning.

Useful Pandas methods:
- `isna()` and `isna().sum()` — detect and count missing values.
- `fillna()` — replace missing values.
- `dropna()` — remove rows with missing values.
- `duplicated()` — identify duplicate rows.
- `drop_duplicates()` — remove duplicate rows.

````python
orders_quality = orders.copy()
orders_quality.loc[2, "customer"] = None

print("Missing values by column:")
print(orders_quality.isna().sum())
````

Do not automatically replace missing values without considering what they mean. For example, `"Unknown"` can be useful for a report, but it does not recover a customer's real identity.

````python
orders_quality["customer"] = orders_quality["customer"].fillna("Unknown")
display(orders_quality.loc[2, ["order_id", "customer"]])
````

### Detect duplicates

The sample intentionally repeats order `AW-4107`. In a real order system, one order may contain multiple product lines, so `order_id` alone is not always a unique line identifier. Here, the repeated row is an accidental duplicate for teaching purposes.

````python
duplicate_ids = orders[orders.duplicated(subset=["order_id"], keep=False)]
display(duplicate_ids[["order_id", "customer", "product", "quantity"]])

orders_deduplicated = orders.drop_duplicates()
print(f"Rows before: {len(orders)}")
print(f"Rows after: {len(orders_deduplicated)}")
````

## 10. Grouping and business summaries

`groupby()` follows a **split–apply–combine** pattern: split rows into groups, calculate a summary for each group, and combine the results.

This is useful for comparing categories, products, and order statuses.

````python
category_summary = (
    orders.groupby("category", as_index=False)
    .agg(
        order_lines=("order_id", "count"),
        units_sold=("quantity", "sum"),
        line_value=("line_total", "sum")
    )
    .sort_values("line_value", ascending=False)
)
display(category_summary)
````

**Interpretation:** This is a summary of order lines, units, and simplified line value by category. The sample includes a duplicate and a cancelled order, so these numbers are for learning only. An official revenue report needs explicit rules for duplicates, cancellations, returns, discounts, and tax.

````python
status_summary = (
    orders.groupby("status", as_index=False)
    .agg(
        order_lines=("order_id", "count"),
        line_value=("line_total", "sum")
    )
    .sort_values("order_lines", ascending=False)
)
display(status_summary)
````

## 11. CSV input and output

CSV is a common text format for tabular data. Pandas uses `to_csv()` to write a DataFrame and `read_csv()` to load a CSV file.

The notebook creates its own local file; no external dataset is needed.

````python
from pathlib import Path

data_folder = Path("adventureworks_data")
data_folder.mkdir(exist_ok=True)
csv_path = data_folder / "orders_unit06.csv"

orders.to_csv(csv_path, index=False)
print(f"Saved: {csv_path}")
````

`index=False` prevents Pandas from writing the DataFrame index as an extra column when it is not part of the business data.

````python
orders_loaded = pd.read_csv(csv_path)

print(f"Loaded shape: {orders_loaded.shape}")
print("Loaded data types:")
print(orders_loaded.dtypes)
display(orders_loaded.head())
````

CSV stores values as text. Pandas infers data types when reading the file, so inspect `dtypes` after loading.

## 13. Instructor answer key

````python
# Reference solution — AdventureWorks data preparation
challenge = orders.copy()

challenge["line_total"] = challenge["quantity"] * challenge["unit_price"]
challenge["value_band"] = np.where(
    challenge["line_total"] >= 100,
    "High line value",
    "Standard line value"
)

processing = challenge[challenge["status"] == "Processing"].copy()

quality_check = challenge.copy()
quality_check.loc[0, "customer"] = None
missing_counts = quality_check.isna().sum()
quality_check["customer"] = quality_check["customer"].fillna("Unknown")

duplicates = challenge[challenge.duplicated(keep=False)]
cleaned = challenge.drop_duplicates().copy()

category_solution = (
    cleaned.groupby("category", as_index=False)
    .agg(units_sold=("quantity", "sum"), line_value=("line_total", "sum"))
    .sort_values("line_value", ascending=False)
)

output_folder = Path("adventureworks_data")
output_folder.mkdir(exist_ok=True)
solution_path = output_folder / "orders_unit06_solution.csv"
cleaned.to_csv(solution_path, index=False)
reloaded = pd.read_csv(solution_path)

print(f"Processing rows: {len(processing)}")
print(f"Duplicate rows flagged: {len(duplicates)}")
print(f"Rows after deduplication: {len(cleaned)}")
print(f"Missing customer values before fill: {missing_counts['customer']}")
print(f"Reloaded shape: {reloaded.shape}")
display(category_solution)
````
