## 2. Lists

A **list** stores an ordered collection of items. Lists are useful when the order matters and the collection may change.

Create a list with square brackets `[]`. A list can contain strings, numbers, Booleans, or even other collections. For now, keep each list focused on one clear purpose.

````python
featured_products = ["Wireless Mouse", "USB-C Hub", "Desk Lamp"]

print(featured_products)
print(f"Number of featured products: {len(featured_products)}")
````

### Updating a list

Lists are **mutable**, which means their contents can be changed after the list is created. You can replace an item, add an item, or remove an item.

````python
featured_products = ["Wireless Mouse", "USB-C Hub", "Desk Lamp"]

featured_products[1] = "Mechanical Keyboard"
featured_products.append("Laptop Stand")
featured_products.remove("Desk Lamp")

print(featured_products)
````

**Expected output**
```text
['Wireless Mouse', 'Mechanical Keyboard', 'Laptop Stand']
```

**Business connection:** A merchandising team may update a featured-products list when a product is promoted or removed.

**Common mistake:** `append()` adds one item to the end. `remove()` removes the first matching value. Both change the existing list.

### Useful list operations

- `len(items)` returns the number of items.
- `items.append(value)` adds one item to the end.
- `items.insert(index, value)` inserts an item at a position.
- `items.remove(value)` removes the first matching value.
- `items.pop()` removes and returns the last item by default.
- `items.sort()` sorts the list in place.

Choose methods based on the task. Do not use `remove()` when you only want to check whether a value exists.

````python
reorder_products = ["USB-C Hub", "Wireless Mouse"]
reorder_products.append("Desk Lamp")
reorder_products.insert(0, "Mechanical Keyboard")

print(reorder_products)
print(f"Items to review: {len(reorder_products)}")
````

## 3. Indexing, slicing, and membership

Lists are ordered, so each item has a position. Python uses zero-based indexing: the first item is at index `0`. Negative indexes count from the end.

````python
products = ["Wireless Mouse", "Mechanical Keyboard", "USB-C Hub", "Desk Lamp"]

print(products[0])   # First item
print(products[1])   # Second item
print(products[-1])  # Last item
````

### Slicing

Slicing creates a new list containing part of the original list. The stop position is excluded.

````python
products = ["Wireless Mouse", "Mechanical Keyboard", "USB-C Hub", "Desk Lamp"]

print(products[0:2])
print(products[:3])
print(products[2:])
````

**Expected output**
```text
['Wireless Mouse', 'Mechanical Keyboard']
['Wireless Mouse', 'Mechanical Keyboard', 'USB-C Hub']
['USB-C Hub', 'Desk Lamp']
```

### Membership

Use `in` to check whether a value exists in a collection. Use `not in` to check that it does not exist.

````python
products = ["Wireless Mouse", "Mechanical Keyboard", "USB-C Hub", "Desk Lamp"]

print("USB-C Hub" in products)
print("Gaming Monitor" in products)
print("Gaming Monitor" not in products)
````

**Why this matters:** Membership checks are useful for validating a product name before processing an order.

**Common mistake:** `products[4]` raises `IndexError` because a four-item list has indexes `0` through `3`.

## 4. Tuples

A **tuple** is an ordered collection that cannot be changed after creation. Tuples are useful for a small group of values that should stay together and remain fixed.

Create a tuple with commas, usually inside parentheses `()`. A tuple with one item needs a trailing comma.

````python
shipping_dimensions_cm = (32, 18, 7)
print(shipping_dimensions_cm)
print(f"Package length: {shipping_dimensions_cm[0]} cm")
print(f"Number of dimensions: {len(shipping_dimensions_cm)}")
````

### Tuple versus list

Use a list when the collection needs to change. Use a tuple when the values represent a fixed group, such as package dimensions for a particular product record.

````python
# A list can be updated
temporary_prices = [24.99, 34.00]
temporary_prices[0] = 22.99

# A tuple is intended to stay fixed
package_dimensions = (32, 18, 7)

print(temporary_prices)
print(package_dimensions)
````

Attempting to assign a new value to a tuple position raises `TypeError`.

## 5. Dictionaries

A **dictionary** stores key-value pairs. It is useful when each value has a clear label, such as an order ID, customer name, or order status.

Create a dictionary with curly braces `{}` and `key: value` pairs. Keys should be unique.

````python
order = {
    "order_id": "AW-1042",
    "customer_name": "Maya Chen",
    "status": "Processing",
    "total": 99.98
}

print(order)
print(order["customer_name"])
print(order["status"])
````

### Reading and updating dictionary values

Access a value using its key. Add a new key or update an existing value using assignment.

````python
order = {
    "order_id": "AW-1042",
    "customer_name": "Maya Chen",
    "status": "Processing",
    "total": 99.98
}

order["status"] = "Shipped"
order["payment_method"] = "Card"

print(order["status"])
print(order["payment_method"])
````

### `get()` and checking keys

Using `order["key"]` raises `KeyError` if the key is missing. The `get()` method can return a default value instead.

````python
order = {"order_id": "AW-1042", "status": "Processing"}

print(order.get("status"))
print(order.get("tracking_number", "Not assigned"))
print("customer_name" in order)
````

**Business connection:** An order has several named attributes. A dictionary lets you access the order status by name rather than remembering a numeric position.

## 6. Sets

A **set** stores unique values. Sets are useful when you need to remove duplicates or compare collections.

Create a set with curly braces or `set()`. An empty set must be created with `set()` because `{}` creates an empty dictionary.

````python
order_categories = {"Accessories", "Home", "Accessories", "Electronics"}
print(order_categories)
print(f"Unique categories: {len(order_categories)}")
````

**Note:** Sets do not guarantee a display order for their elements. Do not rely on their printed order or try to access an item by index.

### Set operations

- `add(value)` adds an item.
- `remove(value)` removes an item and raises an error if it is missing.
- `discard(value)` removes an item if it exists, without raising an error if it is missing.
- `union()` combines unique values.
- `intersection()` finds values shared by both sets.

````python
online_categories = {"Accessories", "Electronics"}
warehouse_categories = {"Home", "Electronics"}

print(online_categories.union(warehouse_categories))
print(online_categories.intersection(warehouse_categories))
````

### Removing duplicates

A set can remove duplicates from a list, but converting back to a list may not preserve the original order. If order matters, do not use a set as a replacement for a list without considering that difference.

````python
raw_categories = ["Home", "Accessories", "Home", "Electronics", "Accessories"]
unique_categories = set(raw_categories)

print(f"Original count: {len(raw_categories)}")
print(f"Unique count: {len(unique_categories)}")
````

## 7. Mutability, immutability, and copying

**Mutable** objects can be changed after creation. Lists, dictionaries, and sets are mutable. Tuples are immutable at the collection level.

A key idea: assigning one variable to another does not automatically create a copy. Both names may refer to the same list.

````python
featured = ["Wireless Mouse", "USB-C Hub"]
featured_alias = featured

featured_alias.append("Desk Lamp")

print(featured)
print(featured_alias)
print(featured is featured_alias)
````

Both variables show the added item because they refer to the same list object.

### Making a shallow copy

For a simple list, `list.copy()` creates a separate list. Updating the copied list does not change the original list's top-level items.

````python
featured = ["Wireless Mouse", "USB-C Hub"]
featured_copy = featured.copy()

featured_copy.append("Desk Lamp")

print(featured)
print(featured_copy)
print(featured is featured_copy)
````

**Important:** A shallow copy does not recursively copy nested mutable objects. We will keep the main example simple here and revisit nested structures next.

## 8. Nested data structures

A nested structure stores one collection inside another. Real business records often contain grouped information: an order can contain customer details and a list of purchased items.

Keep the example small so the structure remains easy to read.

````python
order = {
    "order_id": "AW-2087",
    "customer": {
        "customer_id": "C-104",
        "name": "Lina Haddad"
    },
    "items": [
        {"product_name": "Wireless Mouse", "quantity": 2, "unit_price": 24.99},
        {"product_name": "USB-C Hub", "quantity": 1, "unit_price": 34.00}
    ],
    "status": "Processing"
}

print(order["customer"]["name"])
print(order["items"][0]["product_name"])
print(order["items"][1]["quantity"])
````

````python
# Instructor solution — AdventureWorks promotion
promotion_products = ["Wireless Mouse", "USB-C Hub", "Desk Lamp"]
campaign_dates = ("2026-10-20", "2026-10-27")

campaign = {
    "campaign_id": "PROMO-17",
    "name": "Autumn Essentials",
    "status": "Active"
}

unique_categories = {"Accessories", "Home"}

product_catalogue = [
    {"name": "Wireless Mouse", "price": 24.99, "category": "Accessories"},
    {"name": "USB-C Hub", "price": 34.00, "category": "Accessories"},
    {"name": "Desk Lamp", "price": 28.75, "category": "Home"}
]

premium_products = [
    product["name"]
    for product in product_catalogue
    if product["price"] >= 30
]

has_usb_hub = "USB-C Hub" in promotion_products

print(campaign)
print(f"Campaign dates: {campaign_dates}")
print(f"Unique categories: {unique_categories}")
print(f"Products priced at least $30: {premium_products}")
print(f"USB-C Hub included: {has_usb_hub}")
````
