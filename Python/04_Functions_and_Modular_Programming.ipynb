## 1. Why functions matter

A **function** is a named, reusable block of code. It reduces duplication and gives a business rule one clear place to live.

AdventureWorks may calculate product line totals in several places. Instead of copying the calculation each time, we can define it once and call it whenever needed.

````python
# Repeated calculation: the same logic is written more than once
mouse_total = 24.99 * 2
hub_total = 34.00 * 1

print(f"Mouse line total: ${mouse_total:.2f}")
print(f"USB-C Hub line total: ${hub_total:.2f}")
````

**Discuss:** If the calculation rule changes, why is one reusable function easier to maintain than several copied calculations?

## 2. Defining and calling a function

Use `def` to define a function. Its body must be indented. The body runs when the function is called.

A function name should describe its task. Python style commonly uses lowercase words separated by underscores (`snake_case`).

````python
def greet_customer():
    """Display a welcome message for an AdventureWorks customer."""
    print("Welcome to AdventureWorks!")

greet_customer()
greet_customer()
````

**Expected output**
```text
Welcome to AdventureWorks!
Welcome to AdventureWorks!
```

A function is defined once but can be called many times. The text inside the first function statement is a **docstring**, a short description of the function.

````python
def greet_customer(customer_name):
    """Display a welcome message for a named customer."""
    print(f"Welcome to AdventureWorks, {customer_name}!")

greet_customer("Maya Chen")
greet_customer("Lina Haddad")
````

## 3. Parameters, arguments, and return values

- A **parameter** is a name in a function definition.
- An **argument** is the value supplied when calling the function.
- `return` sends a result back to the calling code.

Use return values when other code needs to calculate with, compare, or store the result.

````python
def calculate_line_total(unit_price, quantity):
    """Return the total price for one product line."""
    return unit_price * quantity

mouse_total = calculate_line_total(24.99, 2)
hub_total = calculate_line_total(34.00, 1)

print(f"Mouse line total: ${mouse_total:.2f}")
print(f"USB-C Hub line total: ${hub_total:.2f}")
````

**Expected output**
```text
Mouse line total: $49.98
USB-C Hub line total: $34.00
```

The function does not need to know the product name. It receives the price and quantity, then returns the calculation.

````python
def calculate_order_totals(unit_price, quantity, discount_rate=0.0):
    """Return subtotal, discount amount, and final total."""
    subtotal = unit_price * quantity
    discount_amount = subtotal * discount_rate
    final_total = subtotal - discount_amount
    return subtotal, discount_amount, final_total

subtotal, discount, final_total = calculate_order_totals(50.00, 2, 0.10)
print(f"Subtotal: ${subtotal:.2f}")
print(f"Discount: ${discount:.2f}")
print(f"Final total: ${final_total:.2f}")
````

## 4. `print()` versus `return`

`print()` displays a value for a person to read. `return` sends a value back to the caller. A function without an explicit `return` returns `None`.

````python
def print_line_total(unit_price, quantity):
    print(unit_price * quantity)

def get_line_total(unit_price, quantity):
    return unit_price * quantity

displayed_result = print_line_total(24.99, 2)
returned_result = get_line_total(24.99, 2)

print(f"Value from print-only function: {displayed_result}")
print(f"Value from return function: {returned_result}")
print(f"Returned value plus $5: ${returned_result + 5:.2f}")
````

**Key point:** Use `return` when the result must be reused in another calculation. Use `print()` when the goal is to display information.

## 5. Positional, keyword, and default arguments

- **Positional arguments** are matched by their order.
- **Keyword arguments** name the parameter explicitly.
- A **default value** makes an argument optional when no value is supplied.

````python
def create_order_summary(order_id, customer_name, status="Processing"):
    """Return a short summary of an order."""
    return f"{order_id} | {customer_name} | {status}"

print(create_order_summary("AW-1042", "Maya Chen"))
print(create_order_summary("AW-1043", "Omar Saleh", "Shipped"))
print(create_order_summary(
    customer_name="Lina Haddad",
    order_id="AW-1044",
    status="Cancelled"
))
````

The first call uses the default status. The second supplies a status positionally. The third uses keyword arguments, so their order in the call can change.

**Rule:** In a function definition, required parameters must come before parameters with default values.

### A default-argument caution

Default values are evaluated when the function is defined. Avoid a mutable list or dictionary as a default unless shared state is intentional. Use `None` and create a new collection inside the function instead.

````python
def add_note(note, notes=None):
    """Return a list containing existing notes and the new note."""
    if notes is None:
        notes = []
    notes.append(note)
    return notes

first_notes = add_note("Confirm payment")
second_notes = add_note("Prepare shipment")
print(first_notes)
print(second_notes)
````

## 6. Scope and variable lifetime

A variable created inside a function is usually **local** to that function. Prefer passing changing values as parameters and returning results instead of modifying global variables.

````python
tax_rate = 0.16  # Example rate for teaching only

def calculate_total_with_tax(price):
    """Return a price after applying the example tax rate."""
    tax_amount = price * tax_rate
    total = price + tax_amount
    return total

print(f"Example total: ${calculate_total_with_tax(100.00):.2f}")
````

`tax_amount` and `total` are local variables. The function reads `tax_rate` from the outer scope. For more reusable code, pass the rate as a parameter when it may change.

````python
def calculate_total_with_rate(price, tax_rate):
    """Return a price after applying a supplied rate."""
    return price * (1 + tax_rate)

print(f"Example total: ${calculate_total_with_rate(100.00, 0.16):.2f}")
````

## 7. Docstrings and readable function design

A **docstring** is the first string inside a function and describes its purpose. Good functions have clear names, one main responsibility, useful parameters, and predictable return values.

````python
def is_reorder_required(stock_quantity, reorder_level):
    """Return True when stock is at or below the reorder level."""
    return stock_quantity <= reorder_level

help(is_reorder_required)
print(is_reorder_required(4, 6))
print(is_reorder_required(18, 8))
````

### Function design checklist
- Use a clear action-oriented name.
- Keep one main responsibility per function.
- Pass changing values as parameters.
- Return results when other code needs them.
- Add a concise docstring.
- Avoid unnecessary global state.
- Follow `snake_case` naming conventions.

Reference: [PEP 8 — Style Guide for Python Code](https://peps.python.org/pep-0008/).

## 8. Validation and simple error handling

Functions should make important assumptions clear. For example, a product price and quantity should not be negative. Here we use a simple validation rule; more detailed exception handling is covered separately.

````python
def calculate_line_total_checked(unit_price, quantity):
    """Return a line total after validating non-negative inputs."""
    if unit_price < 0:
        raise ValueError("unit_price cannot be negative")
    if quantity < 0:
        raise ValueError("quantity cannot be negative")
    return unit_price * quantity

print(f"Line total: ${calculate_line_total_checked(24.99, 2):.2f}")
````

The function raises `ValueError` rather than silently returning a misleading result. The following error example is intentionally commented out so the notebook can run from top to bottom.

## 9. Functions with lists and dictionaries

Functions can accept structured data. An order dictionary may contain a list of items, and a function can calculate the subtotal for any order with that structure.

````python
order = {
    "order_id": "AW-2087",
    "customer": "Lina Haddad",
    "items": [
        {"product_name": "Wireless Mouse", "quantity": 2, "unit_price": 24.99},
        {"product_name": "USB-C Hub", "quantity": 1, "unit_price": 34.00}
    ],
    "status": "Processing"
}

def calculate_order_subtotal(order_record):
    """Return the sum of item totals in an order record."""
    subtotal = 0.0
    for item in order_record["items"]:
        subtotal += item["quantity"] * item["unit_price"]
    return subtotal

print(f"Order {order['order_id']} subtotal: ${calculate_order_subtotal(order):.2f}")
````

**Expected output:** `Order AW-2087 subtotal: $83.98`

The function receives the whole order record and processes each item. The same function can be reused for any order with the same structure.

````python
another_order = {
    "order_id": "AW-2088",
    "customer": "Maya Chen",
    "items": [
        {"product_name": "Desk Lamp", "quantity": 1, "unit_price": 28.75}
    ],
    "status": "Shipped"
}

print(f"Order {another_order['order_id']} subtotal: ${calculate_order_subtotal(another_order):.2f}")
````

## 11. Combining functions into a small workflow

A workflow can be built from several small functions: one calculates the subtotal, one applies a shipping rule, and another creates a readable summary. The thresholds below are invented for this training scenario.

````python
def calculate_order_subtotal(order_record):
    """Return the sum of all item totals in an order record."""
    subtotal = 0.0
    for item in order_record["items"]:
        subtotal += item["quantity"] * item["unit_price"]
    return subtotal

def calculate_shipping_fee(subtotal, free_shipping_threshold=100.00):
    """Return zero for qualifying orders; otherwise return a flat fee."""
    if subtotal >= free_shipping_threshold:
        return 0.0
    return 7.50

def create_order_summary(order_record):
    """Return a readable summary with subtotal, shipping, and total."""
    subtotal = calculate_order_subtotal(order_record)
    shipping_fee = calculate_shipping_fee(subtotal)
    total = subtotal + shipping_fee
    return (
        f"Order: {order_record['order_id']}\n"
        f"Customer: {order_record['customer']}\n"
        f"Subtotal: ${subtotal:.2f}\n"
        f"Shipping: ${shipping_fee:.2f}\n"
        f"Total: ${total:.2f}"
    )

print(create_order_summary(order))
````

**Expected output**
```text
Order: AW-2087
Customer: Lina Haddad
Subtotal: $83.98
Shipping: $7.50
Total: $91.48
```

**Business note:** The shipping fee and threshold are examples created for this course, not claims about a real company's policy.

## 13. Instructor answer key

> **Instructor use:** Allow students to attempt the challenge before revealing the solution.

````python
def calculate_line_total(unit_price, quantity):
    """Return a line total after basic validation."""
    if unit_price < 0 or quantity < 0:
        raise ValueError("Price and quantity must be non-negative.")
    return unit_price * quantity

def is_high_value_order(order_total, threshold=250.00):
    """Return True when an order meets the high-value threshold."""
    return order_total >= threshold

def calculate_order_subtotal(order_record):
    """Return the sum of line totals in an order record."""
    subtotal = 0.0
    for item in order_record["items"]:
        subtotal += calculate_line_total(item["unit_price"], item["quantity"])
    return subtotal

def calculate_shipping_fee(subtotal, free_shipping_threshold=100.00):
    """Return the example shipping fee for an order subtotal."""
    return 0.0 if subtotal >= free_shipping_threshold else 7.50

def create_order_summary(order_record):
    """Return a readable summary for an AdventureWorks order."""
    subtotal = calculate_order_subtotal(order_record)
    shipping_fee = calculate_shipping_fee(subtotal)
    total = subtotal + shipping_fee
    priority = "High value" if is_high_value_order(total) else "Standard"
    return (
        f"Order: {order_record['order_id']}\n"
        f"Customer: {order_record['customer']}\n"
        f"Subtotal: ${subtotal:.2f}\n"
        f"Shipping: ${shipping_fee:.2f}\n"
        f"Total: ${total:.2f}\n"
        f"Priority: {priority}"
    )

sample_orders = [
    {"order_id": "AW-2087", "customer": "Lina Haddad",
     "items": [
         {"product_name": "Wireless Mouse", "quantity": 2, "unit_price": 24.99},
         {"product_name": "USB-C Hub", "quantity": 1, "unit_price": 34.00}]},
    {"order_id": "AW-2088", "customer": "Maya Chen",
     "items": [{"product_name": "Mechanical Keyboard", "quantity": 2, "unit_price": 79.50}]}
]

for sample_order in sample_orders:
    print(create_order_summary(sample_order))
    print("-" * 32)
````
