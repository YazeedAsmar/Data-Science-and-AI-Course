## 2. Variables and data types

A **variable** is a name that refers to a value. Variables make code easier to read and allow information to be reused.

Use meaningful names such as `product_price` rather than unclear names such as `x`. In Python, variable names are case-sensitive and usually use `snake_case`.

````python
product_name = "Wireless Mouse"
product_price = 24.99
stock_quantity = 35
is_available = True

print(product_name)
print(product_price)
print(stock_quantity)
print(is_available)
````

### Common built-in data types

- `str`: text, such as a product name or order status.
- `int`: a whole number, such as a quantity.
- `float`: a number with a decimal part, such as a price.
- `bool`: `True` or `False`, such as whether a product is available.

Use `type()` to inspect a value's type. This is especially helpful when data comes from different sources.

````python
print(type(product_name))
print(type(product_price))
print(type(stock_quantity))
print(type(is_available))
````

**Expected output**
```text
<class 'str'>
<class 'float'>
<class 'int'>
<class 'bool'>
```

**Business connection:** Product names are text, prices are numeric values, quantities are usually whole numbers, and availability can be represented as a Boolean value.

**Common mistake:** `"35"` is a string, not an integer. Although it looks numeric, Python treats it as text until you convert it.

````python
stock_as_text = "35"
print(type(stock_as_text))
````

### Naming and reassignment

A variable can be assigned a new value. Choose names that describe the business meaning of the value.

````python
order_status = "Processing"
print(order_status)

order_status = "Shipped"
print(order_status)
````

## 3. Operators and type conversion

Operators let Python calculate values and compare information.

### Arithmetic operators

The main arithmetic operators are `+`, `-`, `*`, `/`, `//`, `%`, and `**`.

For business calculations, multiplication is often used to calculate a line total: unit price multiplied by quantity.

````python
unit_price = 49.90
quantity = 2
line_total = unit_price * quantity

print(f"Line total: ${line_total:.2f}")
````

### Comparison and logical operators

Comparison operators such as `==`, `!=`, `>`, `<`, `>=`, and `<=` return a Boolean value.

Logical operators `and`, `or`, and `not` combine or reverse conditions.

````python
stock_quantity = 12
reorder_level = 10
is_in_stock = stock_quantity > 0
needs_reorder = stock_quantity <= reorder_level

print("Available:", is_in_stock)
print("Needs reorder:", needs_reorder)
print("Can fulfill order:", is_in_stock and not needs_reorder)
````

**Interpretation:** The product is available, but it does not currently meet the defined reorder condition. In real systems, business rules must be defined carefully; a product can be available and still need replenishment.

### Type conversion

Type conversion changes a value from one type to another. Common conversions include `int()`, `float()`, and `str()`.

Conversion is useful when numeric information arrives as text, such as a value entered in a form.

````python
quantity_text = "3"
unit_price_text = "19.95"

quantity = int(quantity_text)
unit_price = float(unit_price_text)
line_total = quantity * unit_price

print(f"Quantity: {quantity}")
print(f"Line total: ${line_total:.2f}")
````

**Important:** `int("3")` works, but `int("three")` raises `ValueError`. Python cannot convert arbitrary text into a number.

## 4. Strings, indexing, slicing, and string methods

A **string** (`str`) stores text. Product names, order IDs, customer messages, and order statuses are all examples of text.

Strings are ordered sequences of characters. Python uses zero-based indexing: the first character is at index `0`.

````python
product_name = "Wireless Keyboard"

print(product_name[0])   # First character
print(product_name[1])   # Second character
print(product_name[-1])  # Last character
````

**Expected output**
```text
W
i
d
```

### Slicing

Slicing extracts part of a string using `text[start:stop]`. The start index is included, but the stop index is excluded.

````python
product_name = "Wireless Keyboard"

print(product_name[0:8])  # Characters at positions 0 through 7
print(product_name[:8])   # Same result when start is omitted
print(product_name[9:])   # From index 9 to the end
````

### Useful string methods

- `.lower()` and `.upper()` change letter case.
- `.strip()` removes spaces at the beginning and end.
- `.replace(old, new)` replaces part of a string.
- `.title()` capitalizes the first letter of words.
- `.startswith(prefix)` checks how a string begins.

These methods are useful when standardizing text, but do not change the original string in place. They return a new string.

````python
raw_status = "  SHIPPED "
clean_status = raw_status.strip().lower()
product_label = "wireless keyboard".title()
order_code = "AW-1042"

print(clean_status)
print(product_label)
print(order_code.startswith("AW-"))
````

**Business connection:** Order statuses may arrive with inconsistent capitalization or extra spaces. Standardizing text makes comparisons more reliable.

**Common mistake:** `product_name[0:8]` does not include the character at index `8`. The stop index is excluded.

## 5. Input, output, comments, and f-strings

`print()` displays information. `input()` asks a person to enter text and returns that input as a string.

This distinction matters: even if a user types `5`, `input()` returns `"5"`. Convert it before performing numeric calculations.

````python
# This example uses a fixed value so it runs consistently during a demonstration.
quantity_text = "2"
quantity = int(quantity_text)

unit_price = 24.99
total = quantity * unit_price

print(f"Quantity: {quantity}")
print(f"Total: ${total:.2f}")
````

### Comments and f-strings

A comment begins with `#` and explains why code is written or clarifies a decision. Avoid comments that only repeat the code.

An **f-string** starts with `f` and allows expressions or variables inside `{}`. Format specifications such as `:.2f` display a number with two decimal places.

````python
order_id = "AW-1042"
customer_name = "Maya"
order_total = 99.9

# Display a compact order summary for the support team.
print(f"Order {order_id} for {customer_name}: ${order_total:.2f}")
````

## 6. Common errors and debugging

Errors are part of programming. A useful developer reads the error message, identifies the relevant line, and checks the value or syntax that caused the problem.

### Common beginner errors

- `SyntaxError`: Python cannot understand the code's structure.
- `NameError`: a name is not defined in the current runtime.
- `TypeError`: an operation is used with an incompatible type.
- `ValueError`: a function receives a value of the right general type but an invalid format or content.
- `IndexError`: a sequence index is outside the available range.
